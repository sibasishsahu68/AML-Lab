```python
# Chi-square test without scipy, user input version

# Step 1: Take input from user
yes_yes = int(input("Enter number of students who like sports and buy protein powder: "))
yes_no  = int(input("Enter number of students who like sports and do not buy protein powder: "))
no_yes  = int(input("Enter number of students who do not like sports but buy protein powder: "))
no_no   = int(input("Enter number of students who do not like sports and do not buy protein powder: "))

# Step 2: Build observed table
data = [[yes_yes, yes_no],
        [no_yes, no_no]]

# Step 3: Row totals, column totals, grand total
row_totals = [sum(row) for row in data]
col_totals = [sum(col) for col in zip(*data)]
grand_total = sum(row_totals)

# Step 4: Expected frequencies
expected = [[(row_totals[i] * col_totals[j]) / grand_total
             for j in range(len(col_totals))]
            for i in range(len(row_totals))]

# Step 5: Chi-square statistic
chi2 = 0
for i in range(len(data)):
    for j in range(len(data[0])):
        chi2 += (data[i][j] - expected[i][j])**2 / expected[i][j]

# Step 6: Compare with tabulated value
critical_value = 3.841  # df = 1, alpha = 0.05

print("\nChi-square statistic:", chi2)
print("Expected frequencies:", expected)
print("Critical value (0.05, df=1):", critical_value)

if chi2 >= critical_value:
    print("Result: Reject the null hypothesis (variables are dependent).")
else:
    print("Result: Fail to reject the null hypothesis (variables are independent).")
```
