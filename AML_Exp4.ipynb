```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,2,3])
y = np.array([2,3,4])
print("x is =",x)
print("y is =",y)
print("\n")

# Prediction function
def predict(x, w, b):
  return w * x + b

# Cost function (Mean Squared Error)
def cost_function(y_true, y_pred, n_samples):
    return np.sum((y_pred - y_true)**2) / (2 * n_samples)

# Initialize parameters
w = 0
b = 0
print("Initially weight is =",w)
print("Initially bias is =",b)
print("\n")

# Hyperparameters
n = len(x)
alpha = 0.1
epochs = 2 # User requested 2 iterations

cost_history = [] # Initialize cost history

print(f"Number of iterations = {epochs}\n")

for i in range(epochs):
    # Compute current predictions
    y_pred = predict(x, w, b)

    # Compute current cost
    cost = cost_function(y, y_pred, n)
    cost_history.append(cost) # Store the cost

    # Calculate gradients
    dw = (1/n) * np.sum((y_pred - y) * x)
    db = (1/n) * np.sum(y_pred - y)

    # Update parameters
    w = w - alpha * dw
    b = b - alpha * db

    print(f"--- Iteration {i+1} ---")
    print(f"  y prediction = {y_pred}")
    print(f"  Gradient w.r.t. weight (dw) = {dw:.4f}")
    print(f"  Gradient w.r.t. bias (db) = {db:.4f}")
    print(f"  Updated weight (w) = {w:.4f}")
    print(f"  Updated bias (b) = {b:.4f}")
    print(f"  Cost = {cost:.4f}\n")

print("Final parameters after 2 iterations:")
print(f"  Final weight (w) = {w:.4f}")
print(f"  Final bias (b) = {b:.4f}")

#plot the regression line
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='red', label='Actual Data')
plt.plot(x, predict(x, w, b), color='blue', label='Regression Line')
plt.title('Gradient Descent - Linear Regression Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.plot(range(1, epochs + 1), cost_history[:epochs], color='green', marker='o')
plt.title('Cost vs. Iterations')
plt.xlabel('Iteration')
plt.ylabel('Cost (MSE)')
plt.grid(True)
plt.show()
```
